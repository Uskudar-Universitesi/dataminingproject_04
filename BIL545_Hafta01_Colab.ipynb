## 2. İlk deney: iris verisinde karar ağacı ve taban çizgisi

**Konu.** scikit-learn'deki her model aynı kalıpla kullanılır:

1. Veriyi al, **eğitim** ve **test** diye ikiye böl.
2. Modeli seç.
3. `fit`: modeli eğitim verisiyle **öğret**.
4. `predict`: test verisi için **tahmin** et.
5. Tahminleri gerçek etiketlerle karşılaştırıp **değerlendir**.

Bir modelin iyi olup olmadığını tek başına söyleyemeyiz; onu bir **taban çizgisiyle** (baseline) karşılaştırırız. Buradaki taban çizgisi hiçbir şey öğrenmeyen, her zaman en sık sınıfı söyleyen bir "model"dir. Model bunu geçemiyorsa işe yaramıyordur.

**Veri: iris.** 150 çiçek; her birinde 4 ölçüm (çanak ve taç yaprağın boyu ve eni, cm). Hedef: 3 türden hangisi olduğu (her türden 50 çiçek).

🐍 **Python:** `from modül import ad` modülün tamamını değil, yalnız içindeki bir adı açar. Böylece `sklearn.tree.DecisionTreeClassifier` yerine kısaca `DecisionTreeClassifier` yazabiliriz. Beş satırın her biri bir araç getiriyor.

📊 **Konu:** Getirilen araçlar:
* `load_iris`: iris verisini yükleyen işlev (scikit-learn'ün içinde hazır gelir, internet gerekmez).
* `train_test_split`: veriyi rastgele eğitim ve test parçalarına bölen işlev.
* `DummyClassifier`: taban çizgisi; veriye bakmadan basit bir kuralla tahmin yapar.
* `DecisionTreeClassifier`: karar ağacı sınıflandırıcısı (6. haftada ayrıntılı).
* `accuracy_score`: doğruluk = doğru tahmin sayısı / toplam tahmin sayısı.

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
```

🐍 **Python:** `load_iris(return_X_y=True)`: işlevi `return_X_y` adlı ayarı `True` (doğru) yapılarak çağırıyoruz; böylece iki sonuç döner. **Çoklu atama**: eşittirin sağındaki iki sonuç sırayla soldaki iki ada (`X` ve `y`) atanır.

📊 **Konu:** `X` **öznitelik matrisi**dir: her satır bir çiçek, her sütun bir ölçüm (150 × 4). `y` **hedef** (etiket) dizisidir: her çiçeğin türü, 0, 1 ya da 2 diye kodlanmış. Büyük X ve küçük y yazmak bir alışkanlıktır: matris büyük, vektör küçük harfle.

```python
X, y = load_iris(return_X_y=True)
```

🔍 **Ara adım:** `.shape` boyutları verir: (satır, sütun). `X[:3]` ilk 3 satırı gösterir (`:3` = baştan 3. elemana kadar, 3 dâhil değil). `y[-3:]` son üç etiketi gösterir (eksi sayılar sondan sayar). X'in 150 × 4, y'nin 150 elemanlı olduğunu görün.

```python
print(X.shape, y.shape)
print(X[:3])
print(y[:3], y[-3:])
```

🐍 **Python:** Komut iki satıra yayılmış: parantez açık kaldıkça Python alt satırı aynı komutun devamı sayar. İşlev dört sonuç döndürür; çoklu atamayla dört ada atanır. Ayarlar `ad=değer` biçiminde verilir.

📊 **Konu:** Veriyi ikiye bölüyoruz:
* `test_size=0.25`: verinin %25'i (38 çiçek) test için ayrılır, kalan 112 çiçekle model eğitilir. Test verisi modelin hiç görmediği veridir; başarıyı orada ölçeriz.
* `random_state=42`: bölme rastgeledir; bu sayı rastgeleliğin **tohumudur** (seed). Aynı tohum her seferinde aynı bölmeyi verir, böylece sonuç tekrarlanabilir. 42 özel bir sayı değildir.
* `stratify=y`: **tabakalı bölme**: sınıf oranları eğitimde ve testte korunur (her türden yaklaşık eşit sayıda).

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)
```

🔍 **Ara adım:** `np.bincount` her etiketten kaç tane olduğunu sayar: 0'dan kaç, 1'den kaç, 2'den kaç. Testte üç türün sayılarının birbirine yakın olduğunu (12–13) görün: bu `stratify`'ın işidir.

```python
import numpy as np
print(X_train.shape, X_test.shape)
print(np.bincount(y_test))
```

🐍 **Python:** Önce bir model **nesnesi** oluşturulur (`DummyClassifier(...)`), hemen ardından onun `.fit(...)` yöntemi çağrılır. **Yöntem** (method), bir nesneye ait işlevdir ve nokta ile çağrılır. `fit` eğitilmiş modelin kendisini döndürdüğü için sonuç doğrudan `baseline` adına atanabilir.

📊 **Konu:** `strategy="most_frequent"`: eğitim verisindeki **en sık sınıfı** öğrenir ve her çiçek için onu söyler. Öznitelikleri hiç kullanmaz. Üç sınıf eşit sayıda olduğundan doğruluğu yaklaşık 1/3 olmalıdır.

```python
baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
```

🐍 **Python:** Aynı kalıp: nesneyi oluştur, `fit` ile eğit, sonucu bir ada ata.

📊 **Konu:** Karar ağacı, ölçümlere "taç yaprak boyu 2,45 cm'den küçük mü?" gibi sorular sorarak türü bulmayı öğrenir. Ağaç da bazı adımlarda rastgelelik kullanır; `random_state=42` onu da sabitler.

```python
model = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)
```

🔍 **Ara adım:** Kalıbın 4. ve 5. adımını açık açık görelim. `predict` test çiçeklerinin her biri için bir tür tahmin eder. `tahmin == y_test` her eleman için True (doğru) ya da False (yanlış) üretir; `.sum()` True'ları **sayar** (True = 1, False = 0). Doğru sayısını test boyutuna bölmek doğruluğu verir.

```python
tahmin = model.predict(X_test)
print(tahmin[:10])
print(y_test[:10])
print((tahmin == y_test).sum(), "/", len(y_test))
```

🐍 **Python:** İç içe çağrılar **içten dışa** okunur: önce `baseline.predict(X_test)` tahminleri üretir, sonra `accuracy_score(gerçek, tahmin)` doğruluğu hesaplar, sonra `round(..., 3)` üç ondalık basamağa yuvarlar, en son `print` yazar.

📊 **Konu:** `accuracy_score`'a önce **gerçek** etiketler, sonra **tahminler** verilir.

```python
print("Baseline:", round(accuracy_score(y_test, baseline.predict(X_test)), 3))
print("Ağaç    :", round(accuracy_score(y_test, model.predict(X_test)), 3))
```

**Sunumdaki hâli** (tek hücre):

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X, y = load_iris(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y)

baseline = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
model = DecisionTreeClassifier(random_state=42).fit(X_train, y_train)

print("Baseline:", round(accuracy_score(y_test, baseline.predict(X_test)), 3))
print("Ağaç    :", round(accuracy_score(y_test, model.predict(X_test)), 3))
```

✅ **Sonucu okuyalım:** Taban çizgisi **0,316** (38 çiçeğin 12'si: her zaman aynı türü söylemek ancak o türden olanları bilir), ağaç **0,895** (38'in 34'ü). Ağaç taban çizgisini açıkça geçiyor: ölçümlerden gerçekten bir şey öğrenmiş.

Ama dikkat: bu sonuç **tek bir rastgele bölmeye** dayanıyor. Başka bir tohumla 38 test çiçeği değişir ve doğruluk da değişir. Bir sonraki bölüm bunu ölçüyor.

## 3. Tek bölmeye güvenmemek: 30 farklı bölme

**Konu.** Test kümesi küçükse (burada 38 çiçek) tek bir çiçeğin doğru ya da yanlış bilinmesi doğruluğu 1/38 ≈ 0,026 değiştirir. Bu yüzden aynı deneyi **30 farklı rastgele bölmeyle** tekrarlayıp doğruluğun **ortalamasını** ve **yayılımını** (standart sapma, en küçük ve en büyük) raporlarız. 7. haftada bunun sistemli hâlini (çapraz doğrulama) göreceğiz.

🐍 **Python:** `import numpy as np`: numpy modülünü açar ve ona kısa bir takma ad (`np`) verir. Artık `numpy.mean` yerine `np.mean` yazarız. numpy, sayısal hesap için kullanılan dizi (array) kütüphanesidir.

📊 **Konu:** 30 doğruluk değerinin ortalamasını ve standart sapmasını numpy ile hesaplayacağız.

```python
import numpy as np
```

🐍 **Python:** `[]` boş bir **liste**dir. Liste, sıralı bir eleman topluluğudur; sonradan eleman eklenebilir. (Python'un kendisinde "dizi" yoktur; sayısal diziler numpy'dan gelir. Liste ile numpy dizisi farkını 2. ve 3. haftada göreceğiz.)

📊 **Konu:** Her bölmenin doğruluğunu bu listeye ekleyeceğiz.

```python
skorlar = []
```

Şimdi döngü. Önce **tek bir turunu** elle yapalım, sonra döngünün tamamını çalıştıralım.

🐍 **Python:** `for tohum in range(30):` satırı, altındaki **girintili** satırları 30 kez çalıştırır. `range(30)` sırasıyla 0, 1, 2, …, 29 sayılarını verir (0'dan başlar, 30 dâhil değildir); her turda o turun sayısı `tohum` adına atanır. Hangi satırların döngünün içinde olduğunu **girinti** (satır başındaki 4 boşluk) belirler. Satır sonundaki iki nokta (`:`) bir bloğun başladığını gösterir.

🔍 **Ara adım:** Döngünün **ilk turu** (`tohum = 0`). Her turda: (1) o tohumla yeni bir bölme yapılır (`Xa`, `ya` eğitim; `Xt`, `yt` test; kısa adlar sunumda yer kazanmak için), (2) yeni bir ağaç eğitilir, (3) doğruluğu hesaplanır. Ağacın kendi tohumu (42) sabit kalır; değişen yalnız **bölmedir**.

```python
tohum = 0
Xa, Xt, ya, yt = train_test_split(X, y, test_size=0.25, random_state=tohum, stratify=y)
m = DecisionTreeClassifier(random_state=42).fit(Xa, ya)
print(accuracy_score(yt, m.predict(Xt)))
```

🐍 **Python:** Döngünün tamamı. Son satırdaki `skorlar.append(x)`, listenin sonuna `x`'i ekler. Döngü bittiğinde listede 30 sayı olur. (Listeyi burada yeniden boş başlatıyoruz ki hücre ikinci kez çalıştırılırsa 60 sayı birikmesin.)

📊 **Konu:** 30 farklı bölme, 30 ayrı eğitim, 30 doğruluk değeri.

```python
skorlar = []
for tohum in range(30):
    Xa, Xt, ya, yt = train_test_split(
        X, y, test_size=0.25, random_state=tohum, stratify=y)
    m = DecisionTreeClassifier(random_state=42).fit(Xa, ya)
    skorlar.append(accuracy_score(yt, m.predict(Xt)))
```

🔍 **Ara adım:** `len` listenin eleman sayısını verir (30). İkinci satır ilk 10 skoru üç basamağa yuvarlayarak gösterir. Köşeli parantez içindeki `for`, bir listeden yeni bir liste üretmenin kısa yoludur (**liste üreteci**): "skorlar'ın ilk 10 elemanının her biri için round(s, 3)". Skorların bölmeden bölmeye ne kadar değiştiğini görün.

```python
print(len(skorlar))
print([round(s, 3) for s in skorlar[:10]])
```

🐍 **Python:** `f"..."` bir **f-string**dir: metnin içindeki süslü parantezlere (`{}`) yazılan ifade hesaplanıp metne yerleştirilir. `:.3f` biçim belirtecidir: sayıyı ondalık noktadan sonra 3 basamakla yaz. `min` ve `max` Python'un hazır işlevleridir: listedeki en küçük ve en büyük değer.

📊 **Konu:** `np.mean(skorlar)`: 30 doğruluğun **ortalaması**. `np.std(skorlar)`: 30 doğruluğun **standart sapması**, yani bölmeden bölmeye ne kadar oynadığı. (numpy varsayılan olarak n'ye böler; pandas'ın `.std()`'u n − 1'e böler. Küçük farklar bundandır.)

```python
print(f"{np.mean(skorlar):.3f} ± {np.std(skorlar):.3f}")
print(f"{min(skorlar):.3f} – {max(skorlar):.3f}")
```

✅ **Sonucu okuyalım:** Ortalama doğruluk **0,938 ± 0,034**; en kötü bölmede **0,868**, en iyisinde **1,000**. 2. bölümdeki 0,895, bu aralıkta ortalamanın altında kalan bir bölmeydi: şanssız bir bölmeyi modelin gerçek başarısı sanabilirdik.

**Ders:** Bir modelin başarısını **tek bir sayıyla değil, ortalama ve yayılımıyla** raporlayın. Projelerinizde de bu beklenecek.
